## Function for airports (svedavia) & Dates as Parameters

```python
# 1. Get BASE_URL & API KEY from dotenv file 
BASE_URL = os.getenv("BASE_URL")
API_KEY = os.getenv("FLIGHT_KEY")
```

```python
# 2. Create an array with the Svedavia Airports 
SWE_AIRPORTS = {
    "ARN": "Stockholm Arlanda Airport",
    "BMA": "Bromma Stockholm Airport",
    "GOT": "Göteborg Landvetter Airport",
    "MMX": "Malmö Airport",
    "LLA": "Luleå Airport",
    "UME": "Umeå Airport",
    "OSD": "Åre Östersund Airport",
    "VBY": "Visby Airport",
    "RNB": "Ronneby Airport",
    "KRN": "Kiruna Airport",
}
```

**Source used:** https://www.geeksforgeeks.org/python/get-yesterdays-date-using-python/ 

```python
# 3. Create a parameter for the dates to fetch 

# Import date and timedelta class
# from datetime module
from datetime import date
from datetime import timedelta

# Array with the days [today -1, today-2, today-3] 
DATES = [date.today(), date.today() - timedelta(days=1), date.today() -
         timedelta(days=2)]

#print the first day in the array (yesterday)
print(DATES[0])


# print all the dates 
for date in DATES:
    print(date)

```

```python
# 5. Create a header - request - response 
# 6. Loop 
# For each AIRPORT 
# For each DATE

def get_flight_data(direction):
    # Empty list to store the flights in 
    all_flights = []

    headers = {
        "Ocp-Apim-Subscription-Key": API_KEY,
        "Accept": "application/json",
    }
    for airport in SWE_AIRPORTS: 
        for date in DATES: 
            url = f"{BASE_URL}/{airport}/{direction}/{date}"
            response = requests.get(url, headers=headers)
            response.raise_for_status()
            for flight in response.json()["flights"]:
                all_flights.append({**flight, "Swedavia_airport_name": airport, "Date_fetched": str(date)}) # adderade datum för det hämtade datumet också för att få in det i DF 
    return all_flights
```

```python
# 7. Print results 

df_arrivals = pd.json_normalize(get_flight_data("arrivals"), sep="_")
```

```python
df_departures = pd.json_normalize(get_flight_data("departures"), sep="_")
```

## Delay definition

Instead of relying on `remarksEnglish` text (there is no "Delayed" remark for arrivals), compute delay from the timestamps:

- `delay_minutes` = (`actualUtc`, else `estimatedUtc`) - `scheduledUtc`
- `is_delayed` = `delay_minutes` >= 15 (common industry threshold; adjustable in `DELAY_THRESHOLD_MIN`)
- Cancelled (`CAN`) and deleted (`DEL`) flights are flagged separately and have no delay value.
- Flights with no actual or estimated time yet have no delay value (`delay_minutes` is NaN).
- `flight_date` = scheduled time converted to Swedish local time. We use this for "delays by date" instead of `Date_fetched`, because `Date_fetched` is only the date we asked the API for.
- Duplicate flight legs (the same flight listed on two fetched dates, around midnight) are removed before analysis.

Departure column names are assumed to mirror arrivals (`departureTime_*`). The first cell checks this.

```python
# Check which time columns exist in each dataframe
for name, d in [("arrivals", df_arrivals), ("departures", df_departures)]:
    print(name, [c for c in d.columns if "Time" in c])
```

```python
DELAY_THRESHOLD_MIN = 15

def add_delay(df, time_prefix):
    # time_prefix: "arrivalTime" or "departureTime"
    out = df.copy()
    sched = pd.to_datetime(out[f"{time_prefix}_scheduledUtc"], utc=True)
    est = pd.to_datetime(out.get(f"{time_prefix}_estimatedUtc"), utc=True)
    act = pd.to_datetime(out.get(f"{time_prefix}_actualUtc"), utc=True)
    
    # Flight date = scheduled time in Swedish local time (not the date we fetched)
    # Scheduled time never changes, so a delayed flight stays on its planned day
    out["flight_date"] = sched.dt.tz_convert("Europe/Stockholm").dt.date

    out["delay_minutes"] = (act.fillna(est) - sched).dt.total_seconds() / 60
    status = out["locationAndStatus_flightLegStatus"]
    out["is_cancelled"] = status.eq("CAN")
    out["is_deleted"] = status.eq("DEL")
    out["has_time"] = out["delay_minutes"].notna()
    # empty delay_minutes gives False here, so filter on has_time before counting
    out["is_delayed"] = out["delay_minutes"] >= DELAY_THRESHOLD_MIN
    return out

arr = add_delay(df_arrivals, "arrivalTime")
dep = add_delay(df_departures, "departureTime")
```

```python
# Sanity check: how many flights can we classify, and what does the delay distribution look like?
for name, d in [("arrivals", arr), ("departures", dep)]:
    print(f"--- {name}: {len(d)} rows")
    print(d[["is_cancelled", "is_deleted", "has_time", "is_delayed"]].sum())
    print(d["delay_minutes"].describe())
```

```python
# Cross-check against the remarks: the 9 NEGATIVE "Arrives/Estimated" arrivals, are they >= 15 min late?
neg_new_time = arr[arr["remarksEnglish"].apply(
    lambda r: any(m["indicator"] == "NEGATIVE" and m["text"].split()[0] in ("Arrives", "Estimated") for m in r)
)]
neg_new_time[["flightId", "Swedavia_airport_name", "arrivalTime_scheduledUtc", "arrivalTime_estimatedUtc",
              "arrivalTime_actualUtc", "delay_minutes", "is_delayed"]]
```

```python
# Do duplicate flights appear across fetched dates (same leg listed under two Date_fetched values)?
key = ["flightLegIdentifier_flightId", "flightLegIdentifier_flightDepartureDateUtc", "Swedavia_airport_name"]
print("BEFORE - arrivals dup legs:", arr.duplicated(key).sum(), "| departures dup legs:", dep.duplicated(key).sum())

# remove duplicates so each flight is counted only once
arr = arr.drop_duplicates(subset=key)
dep = dep.drop_duplicates(subset=key)

# NEW: check again, both should now be 0
print("AFTER  - arrivals dup legs:", arr.duplicated(key).sum(), "| departures dup legs:", dep.duplicated(key).sum())
```

```python
# Delay rate by airport, airline and date (arrivals; repeat for dep)
# only flights with a time that were NOT cancelled or deleted (10 cancelled/deleted flights still had a time and would otherwise be counted)
flown = arr[arr["has_time"] & ~arr["is_cancelled"] & ~arr["is_deleted"]]

by_airport = flown.groupby("Swedavia_airport_name")["is_delayed"].agg(["sum", "count", "mean"])
by_airline = flown.groupby("airlineOperator_name")["is_delayed"].agg(["sum", "count", "mean"]).sort_values("count", ascending=False)

# group by flight_date (Swedish local date) instead of Date_fetched
by_date = flown.groupby("flight_date")["is_delayed"].agg(["sum", "count", "mean"])

by_airport, by_airline.head(15), by_date
```
